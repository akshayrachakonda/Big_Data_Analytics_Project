```python
# Cell 7 — Example: common EDA metrics
from pyspark.sql.functions import hour, dayofweek, date_format

# We'll try to detect a parsed timestamp column to use for time-based EDA.
ts_cols = [c for c in df.columns if c.endswith("_ts")]
ts_cols
```

```python
# Cell 8 — If we have a timestamp column, do time-based aggregations
if len(ts_cols) > 0:
    ts = ts_cols[0]  # pick first parsed timestamp column
    df_time = df.withColumn("hour_of_day", hour(col(ts))).withColumn("weekday", dayofweek(col(ts)))
    # Rides by hour
    rides_by_hour = df_time.groupBy("hour_of_day").count().orderBy("hour_of_day")
    print("Rides by hour (hour_of_day,count):")
    rides_by_hour.show(24, truncate=False)
    # Convert small result to pandas for plotting later
    rides_by_hour_pd = rides_by_hour.toPandas()
else:
    print("No parsed timestamp column detected — time-based EDA skipped. Columns looked for:", possible_time_cols)
    rides_by_hour_pd = None
```

```python
# Cell 16 — Geospatial clustering (KMeans) if latitude/longitude columns exist
# We'll try to find likely lat/lon columns
lat_candidates = [c for c in df.columns if 'lat' in c.lower() or 'latitude' in c.lower()]
lon_candidates = [c for c in df.columns if 'lon' in c.lower() or 'lng' in c.lower() or 'longitude' in c.lower()]

lat_candidates, lon_candidates
```

```python
# Cell 17 — Prepare and run KMeans (best-effort)
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.clustering import KMeans

if lat_candidates and lon_candidates:
    lat_col = lat_candidates[0]
    lon_col = lon_candidates[0]
    # Drop rows where coords are null and convert to double
    df_coords = df.select(lat_col, lon_col).where(col(lat_col).isNotNull() & col(lon_col).isNotNull())
    df_coords = df_coords.withColumn(lat_col, col(lat_col).cast("double")).withColumn(lon_col, col(lon_col).cast("double"))
    assembler = VectorAssembler(inputCols=[lat_col, lon_col], outputCol="features")
    coords_vector = assembler.transform(df_coords).select("features")
    # Fit KMeans with k=5 (example)
    k = 5
    kmeans = KMeans().setK(k).setSeed(1)
    model = kmeans.fit(coords_vector)
    centers = model.clusterCenters()
    print("Cluster centers (lat, lon):")
    for c in centers:
        print(c)
else:
    print("No latitude/longitude columns detected. Skipping geo clustering.")
```

```python
# Cell 21 — Small ML example: Predict if ride was cancelled (binary) using basic features
from pyspark.ml import Pipeline
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.feature import StringIndexer, VectorAssembler, Imputer
from pyspark.sql.functions import when

if candidates_status:
    status_col = candidates_status[0]

    # Create binary label: 1 if status contains 'cancel', else 0
    df_label = df.withColumn("label", when(col(status_col).rlike("(?i)cancel"), 1).otherwise(0))

    # Choose up to 3 categorical cols and up to 3 numeric cols
    cat_cols = [c for c, t in df_label.dtypes if t == 'string' and c != status_col][:3]
    numeric_cols = [c for c, t in df_label.dtypes if t in ('int', 'double', 'long', 'float')][:3]

    stages = []
    indexed_cols = []

    # Encode categorical features
    for c in cat_cols:
        idx = c + "_idx"
        si = StringIndexer(inputCol=c, outputCol=idx, handleInvalid="keep")
        stages.append(si)
        indexed_cols.append(idx)

    assembler_inputs = indexed_cols + numeric_cols

    if assembler_inputs:
        # Impute numeric columns to remove NaN
        if numeric_cols:
            imputer = Imputer(inputCols=numeric_cols, outputCols=numeric_cols).setStrategy("mean")
            stages.append(imputer)

        # Assemble features
        assembler = VectorAssembler(inputCols=assembler_inputs, outputCol="features", handleInvalid="keep")
        stages.append(assembler)

        # Logistic Regression model
        lr = LogisticRegression(featuresCol="features", labelCol="label", maxIter=20)
        stages.append(lr)

        pipeline = Pipeline(stages=stages)

        # Filter only rows with valid label
        df_model = df_label.where(col("label").isNotNull())

        train, test = df_model.randomSplit([0.8, 0.2], seed=42)

        model = pipeline.fit(train)
        preds = model.transform(test)

        preds.select("label", "prediction", *assembler_inputs).show(10, truncate=False)
    else:
        print("Not enough features (categorical/numeric) to build a toy model.")
else:
    print("No status/cancel column found — skipping classification example.")
```
