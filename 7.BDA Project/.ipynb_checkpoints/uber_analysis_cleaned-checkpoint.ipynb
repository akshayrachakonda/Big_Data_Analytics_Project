```python
# Cell 1 - Imports
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when, count, concat_ws, try_to_timestamp, hour, dayofweek, month
import matplotlib.pyplot as plt

spark = SparkSession.builder.appName("UberAnalysis").getOrCreate()
```

```python

# Load dataset
file_path = "ncr_ride_bookings_first_5000_null.csv"  # make sure file is in working directory
df = spark.read.csv(file_path, header=True, inferSchema=True)

print("Schema:")
df.printSchema()

print("First 5 rows:")
df.show(5, truncate=False)
```

```python
from pyspark.sql.functions import to_timestamp

# Combine Date and Time into a single string, then parse into timestamp
# from pyspark.sql.functions import to_timestamp

df = df.withColumn("datetime", to_timestamp(col("Date"), "yyyy-MM-dd HH:mm:ss"))


# Extract hour, weekday, month safely
df = df.withColumn("hour", hour(col("datetime")))
df = df.withColumn("day_of_week", dayofweek(col("datetime")))
df = df.withColumn("month", month(col("datetime")))

df.select("Date", "Time", "datetime", "hour", "day_of_week", "month").show(5, truncate=False)
```

```python

# Replace nulls in Payment Method with 'Unknown'
df = df.withColumn("Payment Method", when(col("Payment Method").isNull(), "Unknown").otherwise(col("Payment Method")))
```

```python

payment_counts = df.groupBy("Payment Method").agg(count("*").alias("count")).orderBy("count", ascending=False)
pdf = payment_counts.toPandas()

plt.figure(figsize=(8,5))
plt.bar(pdf["Payment Method"], pdf["count"])
plt.xticks(rotation=45)
plt.ylabel("Number of Rides")
plt.title("Distribution of Payment Methods")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("ncr_ride_bookings_first_5000_null.csv")

# Combine Date + Time into a single datetime column
df["datetime"] = pd.to_datetime(df["Date"] + " " + df["Time"], errors="coerce")

# Extract hour
df["hour"] = df["datetime"].dt.hour

# Group by hour to count rides
rides_by_hour = df.groupby("hour").size().reset_index(name="count")

# Plot
plt.figure(figsize=(8,5))
plt.plot(rides_by_hour["hour"], rides_by_hour["count"], marker="o")
plt.xticks(range(0,24))  # show all 24 hours
plt.xlabel("Hour of Day")
plt.ylabel("Number of Rides")
plt.title("Rides by Hour of Day")
plt.grid(True)
plt.show()

```

```python
import matplotlib.pyplot as plt

# Make sure df['datetime'] is parsed correctly
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')

# Extract day of week (0=Monday, 6=Sunday)
df['day_of_week'] = df['datetime'].dt.dayofweek

# Group and count rides per day
rides_by_day = df.groupby("day_of_week").size().reset_index(name="count")

# Map day numbers to names
day_names = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
rides_by_day["day_name"] = rides_by_day["day_of_week"].apply(lambda x: day_names[x])

# Plot
plt.figure(figsize=(8,5))
plt.bar(rides_by_day["day_name"], rides_by_day["count"])
plt.xlabel("Day of Week")
plt.ylabel("Number of Rides")
plt.title("Rides by Day of Week")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Ensure datetime column is parsed
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')

# Extract month number
df['month'] = df['datetime'].dt.month

# Group and count rides per month
rides_by_month = df.groupby("month").size().reset_index(name="count")

# Plot
plt.figure(figsize=(8,5))
plt.bar(rides_by_month["month"], rides_by_month["count"])
plt.xlabel("Month (1=Jan, 12=Dec)")
plt.ylabel("Number of Rides")
plt.title("Rides by Month")
plt.xticks(range(1,13))  # show months 1–12
plt.show()
```

```python
import matplotlib.pyplot as plt

# Group by Booking Status and count
status_counts = df.groupby("Booking Status").size().reset_index(name="count")

# Sort by count descending
status_counts = status_counts.sort_values(by="count", ascending=False)

# Plot
plt.figure(figsize=(8,5))
plt.bar(status_counts["Booking Status"], status_counts["count"])
plt.xticks(rotation=45)
plt.ylabel("Number of Rides")
plt.title("Booking Status Distribution")
plt.show()
```
